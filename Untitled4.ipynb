```python
# Control Structures in Python
# Control structures in Python are used to control the flow of execution of a program.
# They decide which statements should run, how many times they should run, and in what order.
# There are mainly three types of control structures in Python:

# 1. Sequential Control Structure
# 2. Selection/Decision Control Structure
# 3. Loop/Iteration Control Structure

# ====================================================================================================

# 1. Sequential Control Structure
# In a sequential structure, statements are executed one after another in the same order in which they are written.

# Example

name = "Arjun"
age = 20

print(name)
print(age)

# First, name is stored.
# Then, age is stored.
# Finally, both values are printed.
# The statements execute from top to bottom.

# Check the output below
```

```python
# ========================================================================================================

# 2. Selection / Decision Control Structure
# Selection is used when a program needs to make a decision.
# Python mainly uses: if. if-else, if-elif-else

# A. if Statement
# The if statement executes a block only when the condition is true.

age = 20

if age >= 18:
    print("You are an adult")

# Check the output below
# The condition age >= 18 is true, so the print() statement is executed.
```

```python
# B. if-else Statement
# It provides two choices.

marks = 45

if marks >= 40:
    print("Pass")
else:
    print("Fail")

# Check the output below
```

```python
# C. if-elif-else Statement
# It is used when there are multiple conditions.

marks = 75

if marks >= 90:
    print("Grade A+")
elif marks >= 75:
    print("Grade A")
elif marks >= 50:
    print("Grade B")
else:
    print("Fail")

# Check the output below
# Python checks the conditions from top to bottom.
# Since 75 >= 75 is true, it prints Grade A.
```

```python
# ==========================================================================================================
# 3. Loop / Iteration Control Structure
# Loops are used to repeat a block of code multiple times.
# Python has two main loops: for loop, while loop

# A. for Loop: A for loop is commonly used when we know how many times we want to repeat something.

for i in range(1, 6):
    print(i)

# Check the output below
# range(1, 6) generates numbers from 1 to 5.
# The loop prints each number one by one.
```

```python
# B. while Loop: A while loop repeats the code as long as the condition is true.

i = 1

while i <= 5:
    print(i)
    i = i + 1

# Start with i = 1. Check whether i <= 5. Print i.
# Increase i by 1. Stop when i becomes 6.
# Check the output below
```

```python
# ===================================================================================================
# 4. Jump / Transfer Statements
# Python also provides statements that change the normal flow of a loop.
# break: Stops the loop completely.

print("\n output of the break statement")
for i in range(1, 6):
    if i == 3:
        break
    print(i)

# continue: Skips the current iteration and moves to the next one.

print("\n output of the continue statement")
for i in range(1, 6):
    if i == 3:
        continue
    print(i)

# Check the output below
```
