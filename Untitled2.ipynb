```python
# Step 1: Create a dictionary named 'student'
# A dictionary stores data in Key : Value pairs.
#Module 1,  Lecture 1

student = {
    "Name": "Amit",   # Key = Name, Value = Amit
    "Age": 21,        # Key = Age, Value = 21
    "Marks": 85       # Key = Marks, Value = 85
}

# Step 2: Print the complete dictionary
print("Student Dictionary:")
print(student)

# Step 3: Print a blank line for better output formatting
print()

# Step 4: Check the data type of the variable 'student'
print("Data Type of student:")
print(type(student))
```

```python
# Step 1: Create a dictionary named 'college'
# It stores the college name and department details.

college = {
    "name": "ABC Institute",   # College name

    # 'departments' is a list containing dictionaries
    "departments": [
        {
            "name": "CSE",      # First department
            "students": 120     # Number of students in CSE
        },
        {
            "name": "ECE",      # Second department
            "students": 90      # Number of students in ECE
        }
    ]
}

# Step 2: Access the 'departments' list
# [0] means the first department (CSE)
# ["name"] gets the name of that department

print("First Department Name:")
print(college["departments"][0]["name"])
```
